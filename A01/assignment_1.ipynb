```python
import torch
from diffusers import StableDiffusionPipeline
from PIL import Image

#Dreamshaper
device = "cuda" if torch.cuda.is_available() else "cpu"
model_id = "Lykon/dreamshaper-8"

pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16 if device == "cuda" else torch.float32
)
pipe = pipe.to(device)

#Prompt
prompts = [
    "A surrealistic collage in the style of Wistawa Szymborska. A neat, polished alien in a vintage 1950s business suit sits at a retro office desk, typing intensely on a floating typewriter made of jellybeans. Papery collage elements, vintage magazine cutout texture, absurd juxtaposition, vintage color palette, dreamlike logic, highly detailed and witty.",
    "A surrealistic collage in the style of Wistawa Szymborska. A neat, polished alien wearing a retro top hat and round spectacles, sitting at a floating wooden office desk inside a giant jellybean. Swirling paper cutouts, comic texture and cutouts, absurd juxtaposition, dreamlike logic, pop color palette, witty and ironic.",
    "A surrealistic collage in the style of Wistawa Szymborska. A neat, polished alien taking a coffee break at a future cafe water cooler, eating with its hands. Papery collage elements, layered magazine cutouts, absurd juxtaposition, dreamlike surrealism, retro muted colors, humorous tone."
]

negative_prompt = "photorealistic, hyperrealistic, modern digital rendering, 3d render, blurry, low quality, deformed"

#Image
generator = torch.Generator(device).manual_seed(42)

for i, prompt_text in enumerate(prompts, start=1):
    #.txt file
    with open(f"prompt_{i}.txt", "w", encoding="utf-8") as f:
        f.write(prompt_text)

    #image
    image = pipe(
        prompt=prompt_text,
        negative_prompt=negative_prompt,
        generator=generator,
        num_inference_steps=30,
        guidance_scale=7.5
    ).images[0]

    #.jpg file
    image.save(f"image_{i}.jpg")
    print(f"Saved prompt_{i}.txt and image_{i}.jpg")
```
